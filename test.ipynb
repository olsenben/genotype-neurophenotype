```python
import os

data_dir = os.path.join('E:', 'neuro_data', 'processed', 'derivatives')

def find_files_with_extension(root_dir, extension):
    matching_files = []
    for dirpath, dirnames, filenames in os.walk(root_dir):
        for filename in filenames:
            if filename.lower().endswith(extension.lower()):
                full_path = os.path.join(dirpath, filename)
                matching_files.append(full_path)
    return matching_files
file_paths = find_files_with_extension(data_dir, '.fif')
subject_id = os.path.basename(file_paths[0]).split('_')[0]

print(subject_id)
```
